```python
import numpy as np
import pandas as pd
import plotly.express as px
def rename_columns(df):
    df.columns = [col.split('#')[-1].strip() if '#' in col else col.strip() for col in df.columns]
    return df
```

```python
test = rename_columns(pd.read_csv("FieldSweepExamp.csv", comment = ";"))
print(list(test.columns))
px.line(test, x='magnet.field', y="ThetaK", title='PtCo',color = "repeat")
```
