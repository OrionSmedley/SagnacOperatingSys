```python
import numpy as np
import pandas as pd
# import plotly express
import plotly.express as px
def rename_columns(df):
    df.columns = [col.split('#')[-1].strip() if '#' in col else col for col in df.columns]
    return df
```

```python
test = pd.read_csv("initial.csv", comment = ";")
print(test.columns)
# Apply the function to the dataframe
test = rename_columns(test)
print(test.columns)


# px.line(test, x='magnet.field', y=" dem(3)['x'] # X1", title='PtCo')
px.line(test, x='magnet.field', y="ThetaK", title='PtCo')
```
