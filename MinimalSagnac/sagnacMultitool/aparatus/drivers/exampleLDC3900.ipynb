```python
import pyvisa
GPIB_ADDRESS = "GPIB0::2::INSTR"

rm = pyvisa.ResourceManager()
laser = rm.open_resource(GPIB_ADDRESS)

# Print device information to confirm connection
laser.query("*IDN?")
```

```python
laser.write(":LAS:LDI 148")
laser.query(":LAS:LDI?").strip()
laser.write(":TEC:T 21.5")
laser.query(":TEC:T?").strip()
```
