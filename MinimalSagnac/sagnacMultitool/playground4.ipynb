Timeout Testing

```python
import numpy as np
import pandas as pd
from time import sleep, time

timeout = pd.Timestamp.now()
cooldowncountdown = time()

counter = 0
while counter < 100:
    if time() - cooldowncountdown < 40:
        print(counter)
        counter +=1
        sleep(0.05)
    else:
        raise ValueError(f"Took too long to run. Stopped at {timeout}")
```
