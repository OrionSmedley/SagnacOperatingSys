```python
import time
import numpy as np

def safe_ramp_zurich1(voltage):
    v0 = round(myHF2LI.ac1(),3)
    # print(f"Voltage is {v0}")
    if voltage >= v0:
        for i in range(int(len(np.arange(v0,voltage+0.001,0.05)))):
            myHF2LI.ac1(np.arange(v0,voltage+0.001,0.05)[i])
            time.sleep(0.1)
        myHF2LI.ac1(voltage)
    else:
        for i in range(int(len(np.arange(v0,voltage-0.001,-0.05)))):
            myHF2LI.ac1(np.arange(v0,voltage-0.001,-0.05)[i])
            time.sleep(0.1)
            
def safe_ramp_zurich2(voltage):
    v0 = round(myHF2LI.ac2(),3)
    # print(f"Voltage is {v0}")
    if voltage >= v0:
        for i in range(int(len(np.arange(v0,voltage+0.001,0.05)))):
            myHF2LI.ac2(np.arange(v0,voltage+0.001,0.05)[i])
            time.sleep(0.1)
    else:
        for i in range(int(len(np.arange(v0,voltage-0.001,-0.05)))):
            myHF2LI.ac2(np.arange(v0,voltage-0.001,-0.05)[i])
            time.sleep(0.1)
```

```python
import numpy as np
import time
# PID Test
def monitor_temps(trials = 500):
    levels = {0: "off", 1: "low", 2: "medium", 3: "high"}
    res = LK336.query("SETP? 1")
    setp = float(res.replace('+', ''))
    temp, temp_avg, variance = [], 0, 0
    time_0 = time.time()

    for _ in range(trials):
        temp.append(LK336.get_sample_temp())
    temp_avg = np.mean(temp)
    temp_diff = temp[-1] - temp[0] 

    for i in range(len(temp)):
        variance += (temp[i] - temp_avg)**2

    variance = np.sqrt(variance/trials)
    htr_power = float(LK336.query("HTR? 1").replace('+',''))
    mode = levels.get(int(LK336.query('RANGE? 1').strip()), "unknown")
    time_1 = time.time() - time_0 
    heating_speed = round(temp_diff/time_1,5)
    if heating_speed != 0:
        time_remaining = round(abs((setp -LK336.get_sample_temp())/heating_speed))
    else:
        time_remaining = 0

    # print(f"Set point is {setp} K")
    print(f"Current temp is {round(temp_avg,3)}({round(variance,3)}) K at {htr_power}% power (mode: {mode})")
    if time_remaining%60 < 10:
        print(f"Temperature is changing at {round(heating_speed*1000)} mK/s (remaining: {time_remaining//60}:0{time_remaining%60})")
    else: 
        print(f"Temperature is changing at {round(heating_speed*1000)} mK/s (remaining: {time_remaining//60}:{time_remaining%60})")
    
    if abs(heating_speed*1000) > 2 or abs(setp-LK336.get_sample_temp()) > 0.005 or variance > 0.015:
        return False
    else:
        return True
```

```python
status = False
start = time.time()
while status == False:
    status = monitor_temps(250)
end = time.time()

if (end-start)%60 < 10:
    print(f"Finished in {round((end-start)//60)}:0{round((end-start)%60)}")
else:
    print(f"Finished in {round((end-start)//60)}:{round((end-start)%60)}")
```

```python
def ramp_current24(current):
    I0 = keith24.source_current
    start = time.time()
    print(f"Ramping I")
    if current > keith24.source_current:
        while current >  keith24.source_current + 2.5e-5:
            keith24.source_current += 2.5e-5
        keith24.source_current = current
    elif current < keith24.source_current:
        while current <  keith24.source_current - 2.5e-5:
            keith24.source_current -= 2.5e-5
        keith24.source_current = current
    print(f"I set to {keith24.source_current} at {abs(current-I0)/(time.time()-start)} A/s")
    
def ramp_voltage24( voltage):
    V0 = keith24.source_voltage
    start = time.time()
    print(f"Ramping V")
    if  voltage > keith24.source_voltage:
        while  voltage >  keith24.source_voltage + 5e-4:
            keith24.source_voltage += 5e-4
        keith24.source_voltage = voltage
    elif voltage < keith24.source_voltage:
        while voltage <  keith24.source_voltage - 5e-4:
            keith24.source_voltage -= 5e-4
        keith24.source_voltage = voltage
    print(f"V set to {keith24.source_voltage} at {abs(voltage-V0)/(time.time()-start)} V/s")
```
