```python
import time
import numpy as np

def safe_ramp_zurich(voltage):
    v0 = round(myHF2LI.ac(),3)
    print(f"Voltage is {v0}")
    if voltage >= v0:
        for i in range(int(len(np.arange(v0,voltage+0.001,0.05)))):
            myHF2LI.ac(np.arange(v0,voltage+0.001,0.05)[i])
            time.sleep(0.1)
    else:
        for i in range(int(len(np.arange(v0,voltage-0.001,-0.05)))):
            myHF2LI.ac(np.arange(v0,voltage-0.001,-0.05)[i])
            time.sleep(0.1)
```

```python
def ramp_current(current):
    if current > 0.2:
        print("Go to bed. No more scans")
    else:
        I0 = keith24.source_current
        start = time.time()
        print(f"Ramping I")
        if current > keith24.source_current:
            while current > keith24.source_current + 1e-4:
                keith24.source_current += 1e-4
            keith24.source_current = current
        elif current < keith24.source_current:
            while current < keith24.source_current - 1e-4:
                keith24.source_current -= 1e-4
            keith24.source_current = current
        print(f"I set to {keith24.source_current} at {abs(current-I0)/(time.time()-start)} A/s")


ramp_current(0)
```
