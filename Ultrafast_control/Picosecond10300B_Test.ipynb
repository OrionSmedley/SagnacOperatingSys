```python
#For limited commands the syntax is the same for 10300B and 10070A
from PSPL10070A import PSPL10070A
from time import sleep
import time
import winsound

pg = PSPL10070A(8)
```

```python
pg.write('duration 1e-9')
pg.ask('duration?')
```

```python
pg.write('period 1e-5')
pg.ask('period?')
```

```python
for i in range(30):
    winsound.Beep(1000, 100)
    pg.write(f'delay {i}e-9')
    print(pg.ask('delay?'))
    sleep(1)
```

```python
pg.write('amplitude 7v')
pg.ask('amplitude?')
```

```python
pg.ask('offset?')
```

```python
pg.write('trigger EXT')
pg.ask('trigger?')
```

```python
pg.write('level 0.029')
pg.ask('level?')
```

```python
import winsound


for level in range(25, 33):
    for i in range(level-25):
        winsound.Beep(1000, 100)
        time.sleep(0.1)
    
    mv_level = level / 1000  # Convert to volts
    command = f'level {mv_level:.3f}'
    pg.write(command)
    response = pg.ask('level?')
    print(f"Response: {response}")
    time.sleep(3)
```
