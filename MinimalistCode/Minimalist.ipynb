# Fake Aparatus

```python
def set_parameter(**kwargs):
    print( """Simulate setting multiple parameters on the apparatus.""" )
    for name, value in kwargs.items():
        print(f" \t Setting {name} to {value}")

# Test the set_parameter function
test_parameters = {'temperature': 25, 'pressure': 1.0, 'humidity': 50}
set_parameter(**test_parameters)
```

```python
def perform_measurement(current_state={}):
    """Simulate performing a measurement with the current apparatus settings."""
    print("Performing measurement")
    # Simulate data collection
    data = {"measurement_result": 42}  # Placeholder for actual data
    # Combine current_state and data for recording
    record = {**current_state, **data}
    print(f"Recorded data: {record}")

# Test the perform_measurement function
test_state = {'temperature': 25, 'pressure': 1.0, 'humidity': 50}
perform_measurement(test_state)
```

# Helper Functions

```python
# User-provided variables
variables = {
    'Q' : 9,
    'T': 45,
    'B': [-100, 100],
    'B_sweep': [[0, 1, 2], [0, -1, -2]],
    'V': [[-1, 1]]
}
```

```python
def setNpop_topLevel(variables):
    non_list_variables = {k: v for k, v in variables.items() if not isinstance(v, list)}
    set_parameter(**non_list_variables)
    remaining_variables = {k: v for k, v in variables.items() if isinstance(v, list)}
    return remaining_variables

# Example usage

variables2 = {
    'Q' : 9,
    'T': 45
}
variables3 = setNpop_topLevel(variables)
print(variables3)
```

```python
from itertools import product

def cartesian_product(dicts):
    return (dict(zip(dicts, x)) for x in product(*dicts.values()))


# Generate all combinations of variables
combinations = list(cartesian_product(variables3))
print(combinations)
```

# Real Deal

```python
# User-provided variables
variables = {
    'Q' : 9,
    'T': 45,
    'B': [-100, 100],
    'B_sweep': [[0, 1, 2], [0, -1, -2]],
    'V': [[-1, 1]]
}
```

```python
def run_experiment(variables):
    # Step 1
    # set top level params, and remove them from the variables dict
    variables = setNpop_topLevel(variables)
    # Step 2
    if not variables:
        # if there are no variables left, perform the measurement
        perform_measurement()
        return
    else:
        print("There are still variables left to set.")
        # take the caresian product of the remaining variables
        prods = cartesian_product(variables)
        # iterate over the products, recursively setting the parameters and performing the measurement
        for prod in prods:
            run_experiment(prod)

run_experiment(variables)
```
