```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# import hvplot.xarray
# from mpl_toolkits.mplot3d import Axes3D
import matplotlib.cm as cm
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv('2026-06-12/daedalus_rotation_calib_2026-06-12_1.csv', comment = '#')
```

```python
phi, Bx, By, Bz = df['field_azimuth'], df['Xfield_avg'], df['Yfield_avg'], df['Zfield_avg']
```

```python
def cosphi(phi, A, c):
    return A*np.cos(np.deg2rad(phi)) + c

def sinphi(phi, A, c):
    return A*np.sin(np.deg2rad(phi)) + c
```

```python
plt.plot(phi, Bx)
popt, pcov = curve_fit(cosphi, phi, Bx)
plt.plot(phi, cosphi(phi, *popt))
print(popt)
```

```python
plt.plot(phi, By)
popt, pcov = curve_fit(sinphi, phi, By)
plt.plot(phi, sinphi(phi, *popt))
print(popt)
```

```python
plt.plot(phi, Bz)
popt, pcov = curve_fit(sinphi, phi, Bz)
plt.plot(phi, sinphi(phi, *popt))
print(popt)
```
