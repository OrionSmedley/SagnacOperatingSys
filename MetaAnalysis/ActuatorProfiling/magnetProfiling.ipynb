# magnet moves 0.8 seconds per degree

```python
import numpy as np
import pandas as pd
import plotly.express as px
def rename_columns(df):
    df.columns = [col.split('#')[-1].strip() if '#' in col else col.strip() for col in df.columns]
    return df
```

```python
df = rename_columns( pd.read_csv('magnetProfiling.csv', comment=';') )
df["dT"] = df["time.time"].diff()
df["dPhi"] = np.abs(df["magnet.phi"].diff())
px.line(df, x="dPhi", y="dT")
```

```python
from scipy.stats import linregress

fig = px.line(df, x="dPhi", y="dT")
fig.update_layout(xaxis_range=[0, df["dPhi"].max()], yaxis_range=[0, df["dT"].max()])

# Remove NaN values for fitting
df_clean = df.dropna(subset=["dPhi", "dT"])

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(df_clean["dPhi"], df_clean["dT"])

# Report the intercept and slope
print(f"Intercept: {intercept}, Slope: {slope}")

# Add the fitted line to the plot
fig.add_scatter(x=df_clean["dPhi"], y=intercept + slope * df_clean["dPhi"], mode='lines', name='Fit')
fig.show()
```
