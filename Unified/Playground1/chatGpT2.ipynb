```python
def combine_lists(lists, index=0, current=[]):
    if index == len(lists):
        yield current
        return
    for item in lists[index]:
        if isinstance(item, list):
            for sub_item in item:
                yield from combine_lists(lists, index + 1, current + [sub_item])
        else:
            yield from combine_lists(lists, index + 1, current + [item])

# Mock function to simulate experiments
def run_experiment(config):
    print("Running experiment with:", config)

# Main function to setup and run experiments based on nested parameters
def run_experiments_with_nested_lists():
    params = [
        ['a1', 'a2', 'a3'],  # Example parameter A
        ['d1', 'd2'],        # Example parameter D as initialization
        [['b1', 'b2'], ['b3', 'b4']],  # Example parameter B with nested lists
        ["init","measure"]
    ]
    
    for combination in combine_lists(params):
        experiment_config = {
            'paramA': combination[0],
            'paramD': combination[1],
            'paramB': combination[2],
            'paramC': combination[3]
        }
        run_experiment(experiment_config)

# Execute the test
run_experiments_with_nested_lists()
```
