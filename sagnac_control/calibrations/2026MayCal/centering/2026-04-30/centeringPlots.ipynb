```python
import pandas as pd
import plotly.express as px


df1 = pd.read_csv("test_daedalus_center_calib_2026-04-30_1.csv", comment="#", skip_blank_lines=True)
df2 = pd.read_csv("test_daedalus_center_calib_2026-04-30_2.csv", comment="#", skip_blank_lines=True)

df1["elapsed_minutes"] = df1["elapsed_time"] / 60
df2["elapsed_minutes"] = df2["elapsed_time"] / 60

df2.columns
```

```python
px.line(df1, x="elapsed_minutes", y=["X","Y"], title="Center Calibration Data").show()
px.line(df1, x="elapsed_minutes", y="Zfield", title="Center Calibration Data").show()
```

```python
px.line(df2, x="elapsed_minutes", y=["X","Y"], title="Center Calibration Data").show()
px.line(df2, x="elapsed_minutes", y="Zfield", title="Center Calibration Data").show()
```
