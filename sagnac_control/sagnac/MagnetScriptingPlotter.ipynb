## All in one plot

```python
import pandas as pd
import plotly.express as px

# List all keys in the HDF5 file
with pd.HDFStore(file_path, 'r') as store:
    keys = [key for key in store.keys() if 
            key.startswith('/postCalibration/100ohm/') 
            or key.startswith('/postCalibration/330ohm5/') 
            or key.startswith('/postCalibration/281ohm/')
            ]

# Initialize empty DataFrames to concatenate all data
df_a = pd.DataFrame()
df_b = pd.DataFrame()

# Collect data from all keys for both a and b
for experiment in keys:
    dfplt = pd.read_hdf(file_path, key=experiment)
    dfplt['experiment'] = experiment  # Add a column to distinguish between different experiments
    df_a = pd.concat([df_a, dfplt[['f', 'a', 'phi', 'experiment']]], ignore_index=True)
    df_b = pd.concat([df_b, dfplt[['f', 'b', 'phi', 'experiment']]], ignore_index=True)

# filter to phi = -170
df_a = df_a.query('phi == -170')
df_b = df_b.query('phi == -170')


# Plot all a vs f on the same graph
fig1 = px.line(df_a, x='f', y='a', color='experiment', line_dash='phi', log_x=True, title='a vs f for all experiments')
display(fig1)

# Plot all b vs f on the same graph
fig2 = px.line(df_b, x='f', y='b', color='experiment', line_dash='phi', log_x=True, title='b vs f for all experiments')
display(fig2)
```
