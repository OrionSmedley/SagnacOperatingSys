```python
# GENERICS
import os
import time

import numpy as np
import pandas as pd

from matplotlib import pyplot as plt
import plotly.express as px
from IPython.display import display, clear_output
```

```python
file_path = r"C:\Users\Ralph Group\Documents\Data\Orion\calibration\LCRamrTest\test.csv"
dfplt = pd.read_csv(file_path)
```

```python


# Get the unique values of 'f' to determine the number of facets
unique_fs = dfplt['f'].unique()
num_facets = len(unique_fs)

# Create the line plot with dynamic facets
fig = px.line(dfplt, x='phi', y='a', facet_col='f', facet_col_wrap=1)

# Update layout dynamically based on the number of facets
fig.update_layout(
    height=800,
    **{f'yaxis{i+1}': dict(matches=None) for i in range(num_facets)}  # Dynamically create yaxis1 to yaxisN
)

clear_output(wait=True)
display(fig)
```

```python
import pandas as pd
import plotly.express as px

# Read the CSV file
file_path = r"C:\Users\Ralph Group\Documents\Data\Orion\calibration\LCRamrTest\testHighRes.csv"
dfplt = pd.read_csv(file_path)

# Create the line plot
fig = px.line(dfplt, x='f', y=['a', 'b'], log_x=True)

# Update layout to add a secondary y-axis for 'b'
fig.update_traces(yaxis="y2", selector=dict(name='b'))

fig.update_layout(
    yaxis2=dict(title="b", overlaying="y", side="right"),  # Secondary y-axis
    yaxis=dict(title="a"),  # Primary y-axis
    title="Plot with Secondary Y-Axis"
)

display(fig)
```
