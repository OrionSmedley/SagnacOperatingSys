```python
def plotcurve(num, filename = 'C:/Users/Ralph Group/Documents/Data/TiffanyFilms/NiFerritePostPaterning/IV/gating2point2450_80V_run', label = ''):
    df = pd.read_csv(filename + f'{num}.csv')
    plt.plot(df['Voltage'], df['leakage'], label = label)
```

```python
runs = [6, 4, 5, 9, 10, 12,14]
labels = ['80 to -80V day 1', '-80 to 80V day 1', '80 to -80V day 1', '80 to -80V day 2', '-80 to 80V day 3', '80 to -80V day 3 morning','80 to -80V day 3 afternoon']
[plotcurve(run, label = labels[i]) for i, run in enumerate(runs)]
plt.legend()
```
