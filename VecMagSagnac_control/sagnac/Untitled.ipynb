```python
from pymeasure.instruments.keithley import Keithley2400
from time import sleep
```

```python
sourcemeter = Keithley2400(4)
```

```python
current = 0
pulsewidth = 1e-3
sourcemeter.ramp_to_current(current, steps = 4, pause = pulsewidth)
```

```python
current = 0.01
pulsewidth = 1e-3
sourcemeter.ramp_to_current(current, steps = 4, pause = pulsewidth)
sleep(pulsewidth)
sourcemeter.ramp_to_current(0, steps = 4, pause = pulsewidth)
```
